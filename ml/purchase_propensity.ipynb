```
# =========================================
# PURCHASE PROPENSITY MODEL (1 CELL)
# =========================================

import pandas as pd
from snowflake.snowpark.context import get_active_session
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

# Load data
session = get_active_session()

df = session.sql("""
SELECT
    account_code,
    COUNT(*) AS frequency,
    AVG(amount) AS avg_amount,
    CASE WHEN COUNT(*) > 5 THEN 1 ELSE 0 END AS label
FROM ANYCOMPANY_LAB.SILVER.financial_transactions_clean
GROUP BY account_code
""").to_pandas()

```

```
# Features / target
X = df[["frequency", "avg_amount"]]
y = df["label"]

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```

# Model
model = RandomForestClassifier()
model.fit(X_train, y_train)

# Evaluation
y_pred = model.predict(X_test)

```

```
print("=== MODEL PERFORMANCE ===")
print(classification_report(y_test, y_pred))

# Feature importance
print("\nFeature Importance:")
for col, val in zip(X.columns, model.feature_importances_):
    print(col, ":", val)
```
