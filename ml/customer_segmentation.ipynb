```
# =========================================
# 1. IMPORTS
# =========================================
import pandas as pd
from snowflake.snowpark.context import get_active_session
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# =========================================
# 2. LOAD DATA
# =========================================
session = get_active_session()

df = session.table("ANYCOMPANY_LAB.ANALYTICS.CUSTOMER_FEATURES").to_pandas()

df.head()
```

```
# =========================================
# 3. FEATURE PREPARATION
# =========================================
features = df[["FREQUENCY", "AVG_BASKET", "RECENCY"]].fillna(0)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(features)
```

```
# =========================================
# 4. CLUSTERING
# =========================================
kmeans = KMeans(n_clusters=3, random_state=42)
df["segment"] = kmeans.fit_predict(X_scaled)

df.head()
```

```
# =========================================
# 5. ANALYSIS
# =========================================
df.groupby("segment").agg({
    "FREQUENCY": "mean",
    "AVG_BASKET": "mean",
    "RECENCY": "mean"
})
```

```
# =========================================
# 6. BUSINESS INTERPRETATION
# =========================================
print("Segment 0: Clients faibles")
print("Segment 1: Clients fidèles")
print("Segment 2: Clients à fort potentiel")
```
